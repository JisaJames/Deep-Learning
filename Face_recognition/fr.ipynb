```python
import cv2
import numpy as np
import face_recognition as fr
import os
```

```python
path='images'
```

```python
# assigning o/p to a variable
mylist=os.listdir(path)
mylist
```

```python
# i/p and o/p
imgs=[]
classnames=[]
```

```python
for i in mylist:
    imgpath=os.path.join(path,i)
    img1=cv2.imread(imgpath)
    # print(img1)
    imgs.append(img1)
    classnames.append(i.split('.')[0])
print(classnames)
```

```python


# 1.face location
# 2. face encoding
```

```python
# for one img

img1=imgs[0]

cvt_img=cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
face_loc=fr.face_locations(cvt_img)
print(face_loc)

# pip uninstall numpy

# pip install numpy==1.24.0

face_enc=fr.face_encodings(cvt_img,face_loc)[0]
print(face_enc)
```

```python
# creating a func for all images

def faceencodings(images):
    encodelist=[]
    for i in imgs:
        img1=cv2.cvtColor(i,cv2.COLOR_BGR2RGB)
        face_in_frame=fr.face_locations(img1)
        # print(face_in_frame)

        face_encode=fr.face_encodings(img1,face_in_frame)[0]
        encodelist.append(face_encode)
    return encodelist
```

```python
faceencodings(imgs)
```

```python
encodelist_knownface=faceencodings(imgs)
```

```python
video = cv2.VideoCapture(0)

while True:
    suc, img = video.read()

    img1 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    face_loc1 = fr.face_locations(img1)
    # print(face_loc1)

    face_encode1 = fr.face_encodings(img1, face_loc1)
    # print(face_encode1)

    for enc, loc in zip(face_encode1, face_loc1):
        matches = fr.compare_faces(encodelist_knownface, enc)
        # print(matches)

        face_dis = fr.face_distance(encodelist_knownface, enc)
        # print(face_dis)

        match_index = np.argmin(face_dis)
        # print(match_index)

        y1, x2, y2, x1 = loc

        if matches[match_index]:
            name = classnames[match_index]
            color = (0, 255, 0)      # Green for known face
        else:
            name = "Unknown"
            color = (0, 0, 255)      # Red for unknown face

        cv2.rectangle(img, (x1, y1), (x2, y2), color, 3)
        cv2.putText(img, name, (x1, y1 - 10),
                    cv2.FONT_HERSHEY_COMPLEX, 1, color, 2)        

    cv2.imshow("frame", img)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```
