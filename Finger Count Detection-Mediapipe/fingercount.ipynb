```python
import cv2
import mediapipe as mp
```

```python
mph=mp.solutions.hands
mp_draw=mp.solutions.drawing_utils
data=mph.Hands(max_num_hands=1)
```

```python
video =cv2.VideoCapture(0)

while True:

    sucs, img = video.read()

    img1 = cv2.cvtColor(img,cv2. COLOR_BGR2RGB)

    result = data. process(img1)
    
    # print(result.multi_hand_landmarks)
    tipid=[4,8,12,16,20]
    lmlst=[]
    if result.multi_hand_landmarks:
        for x in result.multi_hand_landmarks:
            for id,lm in enumerate(x.landmark):
                # print(id,lm)
                cx=lm.x
                cy=lm.y
                lmlst.append([id,cx,cy])
                if len(lmlst)!=0 and len(lmlst)==21:
                    fingercount=[]

                    if lmlst[12][1]<lmlst[20][1]: #right
                        if lmlst[4][1]>lmlst[3][1]:#closed
                            fingercount.append(0)
                        else:
                            fingercount.append(1)

                    else:#left
                        if lmlst[4][1]<lmlst[3][1]:#closed
                            fingercount.append(0)
                        else:
                            fingercount.append(1)

                    for i in range(1,5): # all fingers except thumb
                        if lmlst[tipid[i]][2]>lmlst[tipid[i]-2][2]:
                            fingercount.append(0)
                        else:
                            fingercount.append(1)
                    # print(fingercount)
                    if len(lmlst)!=0:
                        fingercount1=fingercount.count(1)
                    # print(fingercount1)
                    cv2.putText(img,"count:"+str(fingercount1),(35,150),cv2.FONT_HERSHEY_COMPLEX,3,(0,255,0),3)



                    # 4y>2y, 8y>6y...  ===>closed
                    # lmlst[4][2]>lmlst[4-2][2]
                




            mp_draw.draw_landmarks(img, x, mph.HAND_CONNECTIONS)
            
      

    cv2.imshow ('Frame', img)

    if cv2.waitKey (1) & 0xFF == ord('2'):
        break

video. release ()
cv2.destroyAllWindows()
```
