```python
import ollama
```

```python
from crewai import Agent, Task, Crew, LLM
local_llm=LLM(model="ollama/llama3.2:1b",
              base_url="http://localhost:11434")

```

```python
researcher=Agent(role='Researcher',
                 goal='Identify 3 major trends in {topic}',
                 backstory="you are tech-savy researcher who find most relevant and up-to-date information",
                 llm= local_llm,
                 verbose=True)

```

```python

writer=Agent(role='Writer',
             goal='summarize the research finding into a short report',
             backstory="you are a professional technical witer known for being concise and clear",
             llm=local_llm,
             verbose=True)
```

```python
research_task=Task(description='Search and analyze news about {topic}',
                   expected_output='A bulleted list of the top 3 trends found',
                   agent=researcher)
 
```

```python
writer_task=Task(description='Write a 2 paragraph summary based on researcher finding',
                 expected_output='A clean formatted summary report',
                 agent=writer)
```

```python
crew= Crew(agents=[researcher,writer],
           tasks=[research_task,writer_task],
           memory=False)
```

```python
# result=crew.kickoff(inputs={'topic':'Agentic AI'})
# Execute the Crew (Jupyter Notebook)
result = await crew.kickoff_async(
    inputs={"topic": "Agentic AI"}
)
print("\n\n###############Final Output###########")

print(result)
```
